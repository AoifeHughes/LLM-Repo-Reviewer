# LLM Repo Reviewer Demo

This notebook demonstrates LLM Repo Reviewer with ChromaDB vector storage, tool calling, and GitHub integration capabilities.

```python
# Import the repo reviewer
import sys
sys.path.append('../src')

from LLMRepoReviewer.repo_reviewer import RepoReviewer
```

## Initialize the Repo Reviewer

Configure the reviewer with your OpenAI-compatible API endpoint. This example uses a local LLM server.

```python
# Initialize with local LLM server
reviewer = RepoReviewer(
    api_base_url="http://localhost:11434/v1",  # Ollama or similar
    api_key="sk-dummy",  # Not used by local servers
    embedding_model_name="all-MiniLM-L6-v2",
    chunk_size=1000,
    chunk_overlap=200
)
```

## Analyze a GitHub Repository

Let's analyze a repository directly from GitHub. The reviewer will:
- Clone the repository to a local directory
- Index all files with smart caching
- Create embeddings and store in ChromaDB

```python
# Analyze a GitHub repository
github_url = "https://github.com/fastapi/fastapi"
local_path = reviewer.clone_github_repo(github_url)

stats = reviewer.process_directory(local_path)
print(f"\nAnalysis complete!")
print(f"Total files: {stats['total_files']}")
print(f"Cached files: {stats['cached_files']}")
print(f"Processed files: {stats['processed_files']}")
```
